```sql
CREATE CATALOG IF NOT EXISTS hospital_records_mavena;
USE CATALOG hospital_records_mavena;
CREATE VOLUME IF NOT EXISTS datasets;
```

After running the above code, click the Catalog tab on the left side (three shapes), refresh until you see the "hospital_records_mavena" catalog, click the "hospital_records_mavena" catalog, click the "default" schema", and highlight your cursor over the "datasets" volume. Then, click the three buttons on the right side of the volume and click "Upload to volume". Finally, upload the included datasets into the volume. Run the below code to check if done properly.
